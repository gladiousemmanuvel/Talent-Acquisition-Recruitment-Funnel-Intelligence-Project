```python
import pandas as pd

# Path
path = r"C:\Users\Gladious Rosary E\Downloads\Multi-Company HR Analytics\\"

# Load core datasets
df_apps = pd.read_csv(path + 'applications.csv')
df_candidates = pd.read_csv(path + 'candidates.csv')
df_jobs = pd.read_csv(path + 'jobs.csv')
df_interviews = pd.read_csv(path + 'interviews.csv')
df_offers = pd.read_csv(path + 'offers.csv')

# 1. Convert Date columns to datetime format
df_apps['application_date'] = pd.to_datetime(df_apps['application_date'])

# 2. Merge Applications with Candidates and Jobs
df_merged = df_apps.merge(df_candidates, on='candidate_id', how='left')
df_merged = df_merged.merge(df_jobs, on='job_id', how='left')

print("Merged Data Shape:", df_merged.shape)
print(df_merged.head(2))
```

```python
import pandas as pd
from sqlalchemy import create_engine

# MySQL connection string
# Format: mysql+pymysql://username:password@host:port/database_name
engine = create_engine('mysql+pymysql://root:Rosee@localhost:3306/recruitment_db')

print("Uploading applications_master to MySQL...")
df_merged.to_sql('applications_master', con=engine, if_exists='replace', index=False, chunksize=10000)

print("Uploading interviews to MySQL...")
df_interviews.to_sql('interviews', con=engine, if_exists='replace', index=False, chunksize=10000)

print("Uploading offers to MySQL...")
df_offers.to_sql('offers', con=engine, if_exists='replace', index=False, chunksize=10000)

print("Data successfully exported to MySQL database!")
```
